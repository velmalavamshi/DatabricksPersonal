```python
%sql
create or replace table mde_dev.gold.fact_salesorder
as
select
  sod.SalesOrderID,
  sod.SalesOrderDetailID,
  sod.CarrierTrackingNumber,
  sod.OrderQty,
  sod.ProductID,
  sod.SpecialOfferID,
  sod.UnitPrice,
  sod.UnitPriceDiscount,
  sod.LineTotal,
  soh.RevisionNumber,
  soh.OrderDate,
  soh.DueDate,
  soh.ShipDate,
  soh.Status,
  soh.OnlineOrderFlag,
  soh.SalesOrderNumber,
  soh.PurchaseOrderNumber,
  soh.AccountNumber,
  soh.CustomerID,
  soh.SalesPersonID,
  soh.TerritoryID,
  soh.BillToAddressID,
  soh.ShipToAddressID,
  soh.ShipMethodID,
  soh.CreditCardID,
  soh.CreditCardApprovalCode,
  soh.CurrencyRateID,
  soh.SubTotal,
  soh.TaxAmt,
  soh.Freight,
  soh.TotalDue,
  soh.Comment
from mde_dev.bronze.salesorderdetail sod
inner join mde_dev.bronze.salesorderheader soh 
on sod.salesorderid = soh.salesorderid
```
